```python
import os
from dotenv import load_dotenv

load_dotenv()

client_id = os.getenv("SPOTIFY_CLIENT_ID")
client_secret = os.getenv("SPOTIFY_CLIENT_SECRET")
```

```python
import spotipy
from spotipy.oauth2 import SpotifyOAuth
import pandas as pd
```

```python
sp = spotipy.Spotify(
    auth_manager=SpotifyOAuth(
        client_id=client_id,
        client_secret=client_secret,
        redirect_uri="http://127.0.0.1:8889/callback",
        scope="playlist-read-private"
    )
)
```

```python
token_info = sp.auth_manager.get_cached_token()
print(token_info.keys() if token_info else "No cached token found")
```

```python
refresh_token = token_info['refresh_token']
print(refresh_token is not None)
print(refresh_token)
```

```python
playlist_link = "https://open.spotify.com/playlist/0kg3Niymj7CG4ArhD6lN7W"
```

```python
playlist_URI = playlist_link.split("/")[-1]
```

```python
data = sp.playlist_items(playlist_URI)
data['items'][0]['item']
```

```python
album_list =[]
for row in data['items']:

    album_id = row ['item']['album']['id']
    album_name = row ['item']['album']['name']
    album_release_date = row ['item']['album']['release_date']
    # print(release_date)
    album_total_tracks = row ['item']['album']['total_tracks']
    album_url = row ['item']['album']['external_urls']['spotify']
    # print(album_url)

    album_element = {'album_id': album_id, 'album_name': album_name, 'release_date': album_release_date, 'total_tracks': album_total_tracks, 'url': album_url}

    album_list.append(album_element)
```

```python
data['items'][0]['item']['artists']
```

```python
artist_list = []

for row in data['items']:
    for key, value in row.items():
        if key == "item":
            for artist in value['artists']:
                artist_dict = {
                    'artist_id': artist['id'],
                    'artist_name': artist['name'],
                    'external_url': artist['href']
                }

                artist_list.append(artist_dict)
```

```python
print(artist_list)
```

```python
song_list = []

for row in data['items']:
    song = row['item']

    song_element = {
        'song_id': song['id'],
        'song_name': song['name'],
        'duration_ms': song['duration_ms'],
        'url': song['external_urls']['spotify'],
        'song_added': row['added_at'],
        'album_id': song['album']['id'],
        'artist_id': song['artists'][0]['id']
    }

    song_list.append(song_element)
```

```python
album_df = pd.DataFrame.from_dict(album_list)
```
